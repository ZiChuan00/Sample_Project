```python
import pandas as pd
from pprint import pprint

years = [2021, 2022, 2023]
countries = ['us', 'singapore', 'uk', 'vietnam', 'mauritius', 'canada', 'denmark', 'china']
all_holidays = []

for year in years:
    for country in countries: 
        url = f'https://www.timeanddate.com/holidays/{country}/{year}'
        tables = pd.read_html(url)
        tables = pd.read_html(url, header=0)  # This ensures it uses only the first header
        holiday_table = tables[0]
        holiday_table = holiday_table.iloc[1:].reset_index(drop=True) 
        holiday_table = holiday_table[~holiday_table.iloc[:, 0].astype(str).str.startswith('*')].reset_index(drop=True) 
        holiday_table['Year'] = year
        holiday_table['Country'] = country 
        all_holidays.append(holiday_table)

df = pd.concat(all_holidays, ignore_index=True) 
rows_to_drop = []

for idx, value in enumerate(df.iloc[:, 0]):  # iterate over first column
    if pd.isna(value):  # check if NaN
        rows_to_drop.append(idx)

# Drop rows with NaN in first column
df = df.drop(rows_to_drop).reset_index(drop=True) 
df = df.rename(columns={'Unnamed: 1': 'Weekday'}) #rename of columns name
print (df)
```

```python
# Make sure 'Date' is a string before concatenation
df['Date'] = df['Date'].astype(str).str.strip()

# Convert the 'Year' column as a string
df['Year'] = df['Year'].astype(str).str.strip()

# Combine date and year, then parse as datetime
df['FullDate'] = pd.to_datetime(df['Date'] + ' ' + df['Year'])

# Drop unnecessary columns and rename
df.drop(columns=['Date', 'Year'], inplace=True)
df.rename(columns={'FullDate': 'Date'}, inplace=True)

#save file to CSV
df.to_csv('holidays_2021_2023.csv', index=False)
print("US holidays data saved successfully!")
```
